### Question 1: Calculate the product of first 10 natural numbers

```python
print("Question 1: Calculate the product of first 10 natural numbers")
product = 1
for num in range(1, 11):
    product *= num
print("Product of first 10 natural numbers:", product)
```

 ### Question 5: Calculate the sum of digits in number 12345

```python
n = 12345
digit = 0
while n > 0:
    digit += n % 10 
    n //= 10
print("The sum of the digits of 12345 is", digit)
```

### Question 6: Check if 97 is a prime number

```python
n = 97

if n>1:
    for i in range(2, int(n/2)+1):
        if (n % i) == 0:
            print(n, "is not a prime number")
            break
    else:
        print(n, "is a prime number")
```

### Question 7: Find the factorial of 8

```python
n = 8
fact = 1
while n > 1:
    fact *= n
    n -= 1
print("\nQuestion 7: Calculate the factorial of 8")
print("The factorial of 8 is", fact)
```

### Question 8: Calculate the average of numbers: 15, 23, 31, 42, 56

```python
numbers_lst = [15, 23, 31, 42, 56]
average = sum(numbers_lst) / len(numbers_lst)
print("\nQuestion 8: Calculate the average of the numbers in the list [15, 23, 31, 42, 56]")
print("The average of the numbers in the list is", average)
```

### Question 9: Find the greatest common divisor (GCD) of 48 and 36

```python
import math
gcd_value = math.gcd(48, 36)
print("\nQuestion 9: Calculate the GCD of 48 and 36")
print("The GCD of 48 and 36 is", gcd_value)
```

### Question 10: Calculate the sum of first 20 odd numbers

```python
n = 0
for i in range(1, 21):
    if i % 2 != 0:
        n += i
print("Sum of odd numbers from 1 to 20 is", n)
```
