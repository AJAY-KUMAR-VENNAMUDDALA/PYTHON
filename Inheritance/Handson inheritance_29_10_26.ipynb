# Practice Question: Create an ATM Class in Python

## Problem Statement

Create a Python class named `ATM` to simulate basic ATM operations.

### Requirements

**1. Class and Attributes**

- Create a class named `ATM`.
- Initialize the class with a `balance` attribute representing the amount of money available in the account.

**2. Deposit Money**

- Create a method named `deposit(amount)`.
- Add the deposited amount to the current balance.
- Display a success message along with the updated balance.
- Do not allow zero or negative deposit amounts.

**3. Withdraw Money**

- Create a method named `withdraw(amount)`.
- Check whether the withdrawal amount is positive.
- If the withdrawal amount is less than or equal to the available balance, deduct it from the balance.
- If the withdrawal amount exceeds the available balance, display an `Insufficient balance` message without changing the balance.
- Do not allow zero or negative withdrawal amounts.

**4. Check Balance**

- Create a method named `check_balance()` that displays the current account balance.

**5. Create an Object**

- Create an object of the `ATM` class with an initial balance of ₹5,000.
- Perform deposit, withdrawal, and balance-checking operations.

### Expected Output

For example, if you deposit ₹2,000, withdraw ₹3,000, and then check the balance:

```text
Initial Balance: ₹5000

₹2000 deposited successfully.
Current Balance: ₹7000

₹3000 withdrawn successfully.
Current Balance: ₹4000

Available Balance: ₹4000
```

If you attempt to withdraw ₹6,000 when the balance is ₹4,000:

```text
Insufficient balance!
Available Balance: ₹4000
```

### Additional Test Cases

| Operation | Amount | Expected Result |
|---|---:|---|
| Deposit | ₹2,000 | Balance increases |
| Withdraw | ₹1,500 | Balance decreases if sufficient |
| Withdraw | ₹10,000 | Insufficient balance message |
| Deposit | ₹0 | Invalid deposit message |
| Withdraw | -₹500 | Invalid withdrawal message |
| Check balance | — | Displays current balance |

### Your Task

Write the complete Python program using a class, `__init__()`, `self`, and instance methods.

**Note:** Do not use standalone functions for the ATM operations. Implement all operations as methods inside the `ATM` class.

````python
class Atm:
    def __init__(self,balance):
        self.balance = balance
        print(f"Balance : {self.balance}")
        
    def deposit(self,amount):
        if amount<=0:
            print("Invalid Deposit amount!!")
        else:
            self.balance +=amount
            print(f"{amount} is deposited successfully")
            print(f"Current Balance :{self.balance} ")
            
    def withdraw(self,amount):
        if amount > self.balance :
            print("Insufficient Balance!!!")
        elif amount <= self.balance:
            self.balance -= amount
            print(f"{amount} withdrawn successfully")
            print(f"Current balance :{self.balance}")
        else:
            print("Invalid amount for withdrawl")
    def check_balance(self):
        print(f"Avaliable balance is {self.balance}")
````

````python
atm = Atm(6000)

atm.deposit(500)
atm.withdraw(68000)
atm.withdraw(300)
atm.check_balance()
````

## Problem Statement

Create a Python program to demonstrate **inheritance** using a parent class named `Animal` and two child classes named `Dog` and `Cat`.

Write the complete Python program using:

### Important Note

Do not define the `eat()` method separately inside the `Dog` and `Cat` classes. Both child classes must inherit this method from the `Animal` parent class.

This exercise demonstrates **single inheritance and code reusability** in Python.

````python
class Animal:
    def __init__(self,name, age):
        self.name = name
        self.age = age
    def eat(self):
        print(f" {self.name} is eating")
        
    def display_info(self):
        print(f"{self.name} ")
    
````
