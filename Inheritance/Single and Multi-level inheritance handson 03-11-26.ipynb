### Examples of Single - level Inheritance


```python
#  Example 1: Bank deposit and Savigs accounts with interest rate

class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        
        print(f"Deposited {amount}. New balance: {self.balance}")
        

class SavingsAccount(BankAccount):
    def __init__(self, balance, interest_rate):
        super().__init__(balance)
        self.interest_rate = interest_rate
    
    def apply_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        
        print(f"Applied interest of {interest}. New balance: {self.balance}")
        
bank = SavingsAccount(1000, 0.05)
bank.deposit(500)
bank.apply_interest()
```

```python
# Example 2: Student and GraduateStudent classes
class Student: 
    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name
        
    def display_info(self):
        print(f"Student ID: {self.student_id}, \nName: {self.name}")
        
class GraduateStudent(Student):
    def __init__(self, student_id, name, branch):
        super().__init__(student_id, name)
        self.branch = branch
        
    def show_branch(self):
        super().display_info()
        print(f"Branch: {self.branch}")
        
student = GraduateStudent("G123", "Ajay","computer science")
student.show_branch()
    
```

## Example of multi level Inheritance

```python
## Example 1: Manager and Employee classes
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        
    def display_info(self):
        print(f"Employee Name: {self.name}, \nSalary: {self.salary}")
class Manager(Employee):
    def __init__(self,name,salary,department):
        super().__init__(name,salary)
        self.department = department
        
    def display_info1(self):
        super().display_info()
        print(f"Manager Name: {self.name}, Salary: {self.salary}, Department: {self.department}")
        
class ProductManager(Manager):
    def __init__(self,name,salary,department,product):
        super().__init__(name,salary,department)
        self.product = product
        
    def display_info(self):
        super().display_info1()
        print(f"Product Manager Name: {self.name}, Salary: {self.salary}, Department: {self.department}, Product: {self.product}")


obj = ProductManager("Ajay", 50000, "IT", "Software")
obj.display_info()
        
```

```python
## Example 2. Student and teacher classes

class Person: 
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")
        
class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id
        
    def display_info(self):
        super().display_info()
        print(f"Student ID: {self.student_id}")
        
        
class Teacher(Student):
    def __init__(self, name, age, student_id, subject):
        super().__init__(name, age, student_id)
        self.subject = subject

    def display_info(self):
        super().display_info()
        print(f"Subject: {self.subject}")
        

p1 = Person("Ajay", 25)


s1 = Student("Ajay", 25, "S123")

t1 = Teacher("Lohit", 25, "S123", "Mathematics")
t1.display_info()
```
