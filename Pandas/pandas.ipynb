```python
import pandas as pd
import numpy as np

def optimize_memory(df):
    before = df.memory_usage(deep = True).sum() / 1024**2
    print(f"Memory Before: {before : .4f} MB")
    
    for col in df.select_dtypes(include=["int"]).columns:
        col_min = df[col].min()
        col_max = df[col].max()
        old_type = df[col].dtype
        
        if col_min >= np.iinfo(np.int8).min and col_max <= np.iinfo(np.int8).max:
            df[col] = df[col].astype(np.int8)
            
        elif col_min >= np.iinfo(np.int16).min and col_max <= np.iinfo(np.int16).max:
            df[col] = df[col].astype(np.int16)
            
        elif col_min >= np.iinfo(np.int32).min and col_max <= np.iinfo(np.int32).max:
            df[col] = df[col].astype(np.int32)
            
        elif col_min >= np.iinfo(np.int64).min and col_max <= np.iinfo(np.int64).max:
                df[col] = df[col].astype(np.int64)
                
                print(f"{col} : {old_type} -> {df[col].dtype}")
            
        
    after = df.memory_usage(deep=True).sum() / 1024**2

    lift = ((before - after) / before * 100) if before else 0

    print(f"\nMemory After : {after:.4f} MB")
    print(f"Memory Saved : {before - after:.4f} MB")
    print(f"Memory Reduction: {lift:.2f}%")

    return df
```

```python
df = optimize_memory(df)
```

```python
import pandas as pd
import numpy as np

def treat_null_values(df, threshold=30):

    print("Null Values Before Treatment:")
    print(df.isnull().sum())

    for col in df.columns:

        null_percentage = df[col].isnull().mean() * 100

        # Drop column if null percentage > 30%
        if null_percentage > threshold:
            df.drop(columns=[col], inplace=True)
            print(f"{col}: Dropped ({null_percentage:.2f}% nulls)")

        # Numerical columns - Median
        elif pd.api.types.is_numeric_dtype(df[col]):

            median_value = df[col].median()
            df[col] = df[col].fillna(median_value)

            print(f"{col}: Filled with Median = {median_value}")

        # Categorical columns - Mode
        elif pd.api.types.is_object_dtype(df[col]) or isinstance(
            df[col].dtype, pd.StringDtype
        ) or isinstance(df[col].dtype, pd.CategoricalDtype):

            mode_value = df[col].mode(dropna=True)

            if not mode_value.empty:
                fill_value = mode_value.iloc[0]
                df[col] = df[col].fillna(fill_value)

                print(f"{col}: Filled with Mode = {fill_value}")

    print("\nNull Values After Treatment:")
    print(df.isnull().sum())

    return df


# Usage
df = treat_null_values(df, threshold=30)
```
