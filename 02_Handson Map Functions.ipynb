## Example 1: Double numbers

```python
print(list(map(lambda x:x*2,[1,2,3,4])))
```

## Example - 2: Convert strings to numbers

```python
print(list(map(int, ["10","20","30","40"])))
```

## Example - 3: Upper case names

```python
print(list(map(str.upper,["asha","ravi","meena"])))
```

## Example - 4: Map with named function

```python
def celsius_to_function(c):
    return c * 9/5 +32
print(list(map(celsius_to_function,[0,20,37,100])))
```

## Example - 5: Map two iterables

```python
print(list(map(lambda a,b:a+b,[1,2,3],[10,20,30])))
```

## Example - 6: Clean user inputs

```python
raw = ["  alice  ","  BOB ", "  Chitra "]
print(list(map(lambda s:s.strip().title(),raw)))
```

## Example - 7: Map dictionaries

```python
products=[{"name":"Pen","price":20},{"name":"Bag","price":800}]
print(list(map(lambda p:{**p, "price_with_tax":round(p["price"]*1.18,2)},products)))
```

## Example 8: Map with conditional pricing

```python
prices = [500, 1000, 1500, 2000]
print(list(map(lambda p:p*0.9 if p>=1000 else p,prices )))
```

## Example 9: Map with multiple fields

```python
quantites = [2,3,5]
unit_prices = [100, 80 ,50]
print(list(map(lambda q,p :q*p, quantites,unit_prices)))
```

## Example 10 : Complex retail order enrichment

```python
orders=[{"id":1,"qty":3,"unit_price":250,"tier":"gold"},{"id":2,"qty":2,"unit_price":800,"tier":"regular"}]
def enrich(o):
    subtotal = o["qty"]*o["unit_price"]
    discount = 0.15 if o["tier"] == "gold" else 0.05
    return {**o, "subtotal":subtotal,"net":round(subtotal * (1 -discount),2)}
print(list(map(enrich,orders)))
```
