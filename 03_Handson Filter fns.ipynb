## Ex - 1: Even numbers

```python
print(list(filter(lambda x:x%2 == 0, range(1,11))))
```

## Ex - 2: Positive Values 

```python
print(list(filter(lambda x:x>0, [-5,3,0,-2,8])))
```

## Ex - 3 : Non-Empty strings


```python
print(list(filter(None, ["Python","", "SQL",None,"Azure"])))
```

## Ex - 4: Names starting with A

```python
print(list(filter(lambda s:s.lower().startswith("a"),["Asha","Ravi","Anil","Meena"])))
```

## Example - 5: Marks above cuttoff

```python
marks = [35,67,82,40,91]
print(list(filter(lambda c:c>=60, marks)))
```

## Example - 6: Multiple conditions

```python
nums = range (1,51)
print(list(filter( lambda x:x%3 == 0 and x % 2 == 0, nums)))
```

## Example - 7: Filter list of dictionaries

```python
staff=[{"name":"A","active":True},{"name":"B","active":False},{"name":"C","active":True}]
print(list(filter(lambda e:e["active"],staff)))
```

## Example 8: Filter invalid records

```python
rows=[{"id":1,"amount":200},{"id":2,"amount":-5},{"id":None,"amount":90},{"id":4,"amount":0}]
print(list(filter(lambda m:m["id"] is not None and m["amount"] > 0,rows)))
```

## Example 9: Churn -risk customers


```python
customers=[{"name":"A","churn_probability":0.83,"spend":2000},{"name":"B","churn_probability":0.31,"spend":4000},{"name":"C","churn_probability":0.91,"spend":700}]
print(list(filter(lambda x:x["churn_probability"] >= 0.8 and x["spend"] >= 1000, customers)))
```

## Example 10:  Complex order qualification


```python
orders=[{"id":1,"status":"paid","amount":2500,"country":"IN"},{"id":2,"status":"pending","amount":3000,"country":"IN"},{"id":3,"status":"paid","amount":500,"country":"US"}]
def eligible(o):
    return o["status"] == "paid" and o["amount"] >= 1000 and o["country"] == "IN"
print(list(filter(eligible, orders)))
```
