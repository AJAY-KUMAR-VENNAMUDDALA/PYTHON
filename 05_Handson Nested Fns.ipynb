## Eg - 1: Simple Nested function

```python
def outer():
    def inner():
        return "Hello from inner"
    return inner()
print(outer())
    
```

## Eg - 2: Inner Uses outer variable


```python
def greet(name):
    def message():
        return f"Welcome, {name}!"
    return message()
print(greet("Asha"))
```

## Eg - 3: Closure factory

```python
def multiplier(factor):
    def multiply(value):
        return factor*value
    return multiply
triple=multiplier(3)
print(triple(7))
```

## Eg - 4: Nested Validation

```python
def register_user(name,age):
    def valid():
        return bool(name.strip()) and age>=18
    return "Registered" if valid() else "Rejected"
print(register_user("Ravi",22))
```

## Eg - 5: Non-local counter

```python
def make_counter():
    count=0
    def increment():
        nonlocal count
        count+=1
        return count
    return increment
c=make_counter()
print(c(),c(),c())
```

## Eg - 6: Nested Function with default

```python
def tax_calculator(rate):
    def total(price,quantity=1):
        return round(price*quantity*(1+rate),2)
    return total
gst18=tax_calculator(0.18)
print(gst18(200,3))
```

## Eg -7 : Decorator basics

```python
def log_calls(fn):
    def wrapper(*args,**kwargs):
        print("Calling",fn.__name__)
        return fn(*args,**kwargs)
    return wrapper
@log_calls
def add(a,b):
    return a+b
print(add(4,5))
```

## Eg - 8: Function returning function

```python
def make_threshold_filter(limit):
    def check(value):
        return value>=limit
    return check
print(list(filter(make_threshold_filter(70),[50,75,90,60])))
```

## Eg - 9 : Nested retail discount rules

```python
def pricing_engine(tier):
    discounts={"regular":0.05,"silver":0.1,"gold":0.2}
    def price(amount):
        discount=discounts.get(tier,0)
        return round(amount*(1-discount),2)
    return price
gold_price=pricing_engine("gold")
print([gold_price(p) for p in [100,500,1000]])
```

## Eg - 10: Complex decorator with wraps

```python
from functools import wraps
def require_positive(fn):
    @wraps(fn)
    def wrapper(*args,**kwargs):
        if any(isinstance(v,(int,float)) and v<0 for v in args):
            raise ValueError("Negative numeric argument")
        return fn(*args,**kwargs)
    return wrapper
@require_positive
def invoice(unit_price,quantity):
    return unit_price*quantity
print(invoice(250,4),invoice.__name__)
```
