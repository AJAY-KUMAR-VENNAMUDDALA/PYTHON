## Eg - 1: Sum numbers

```python
from functools import reduce
print(reduce(lambda a,b: a+b,list(range(1,6))))
```

## Eg - 2: Multiply Numbers


```python
from functools import reduce
print(reduce(lambda a,b: a*b,list(range(1,5))))
```

## Eg - 3: Maximum Number

```python
from functools import reduce
print(reduce(lambda a,b:a if a>b else b, [8,2,19,4]))
```

## Eg - 4: Concatenation of words

```python
print(reduce(lambda a,b:a+" "+b,["Python","is","powerfull"]))
```

## Eg - 5: Use Initializer

```python
print(reduce(lambda acc, x: acc+x,[],100))
```

## Eg -6: Count Occurences

```python
items = ["apple","banana","apple","orange","banana","apple"]
def count(acc, item):
    acc[item]= acc.get(item,0)+1
    return acc
print(reduce(count,items,{}))
```

## Eg - 7: Flatten nested lists

```python
print(reduce(lambda a,b:a+b,[[1,2],[3],[4,5]],[]))
```

## Eg - 8: Total Order revenue

```python

orders=[{"qty":2,"price":100},{"qty":3,"price":250},{"qty":1,"price":500}]
print(reduce(lambda total,o:total+o["qty"] * o["price"],orders,0))
```

## Eg - 9: Group Revenue by category

```python
orders=[{"category":"Books","amount":300},{"category":"Tech","amount":2000},{"category":"Books","amount":150}]
def aggregate(acc,o):
    key=o["category"]
    acc[key]=acc.get(key,0)+o["amount"]
    return acc
print(reduce(aggregate,orders,{}))
```

## Eg - 10: Complex daily sales summary


```python
sales=[{"store":"A","amount":100,"returned":False},{"store":"B","amount":300,"returned":False},{"store":"A","amount":80,"returned":True},{"store":"A","amount":150,"returned":False}]
def summarize(acc,s):
    store=s["store"]
    acc.setdefault(store,{"revenue":0,"valid_orders":0,"returns":0})
    if s["returned"]:
        acc[store]["returns"]+=1
    else:
        acc[store]["revenue"]+=s["amount"]
        acc[store]["valid_orders"]+=1
    return acc
print(reduce(summarize,sales,{}))
```
