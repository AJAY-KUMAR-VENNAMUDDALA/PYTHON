## Eg - 1: Count Down

```python
def countdown(n):
    if n<=0:
        return []
    return [n]+countdown(n-1)
print(countdown(5))
```

## Eg - 2: Factorial

```python
def factorial(n):
    if n<0: raise ValueError("n must be non negative")
    if n<=1: return 1
    return n*factorial(n-1)
print(factorial(5))
```

## Eg - 3: Sum of First N


```python
def sum_n(n):
    if n<=0:return 0
    return n+sum_n(n-1)
print(sum_n(10))
```

## Eg - 4: Fibbonacci with memorization

```python
from functools import lru_cache
@lru_cache(None)
def fibonacci(n):
    if n<0:raise ValueError("n must be nonnegative")
    if n<2:return n
    return fibonacci(n-1)+fibonacci(n-2)
print([fibonacci(n) for n in range(12)])
```

## Example 5: reverse String


```python
def reverse_text(s):
    if len(s)<=1:return s
    return reverse_text(s[1:])+s[0]
print(reverse_text("python"))
```

## Example - 6: Digit sum

```python
def digit_sum(n):
    n=abs(n)
    if n<10:return n
    return n%10+digit_sum(n//10)
print(digit_sum(98765))
```

## Eg - 7: GCD

```python
def gcd(a,b):
    if b==0:return abs(a)
    return gcd(b,a%b)
print(gcd(48,18))
```

## Eg 8: Recursive binary search

```python
def binary_search(values,target,lo=0,hi=None):
    if hi is None:hi=len(values)-1
    if lo>hi:return -1
    mid=(lo+hi)//2
    if values[mid]==target:return mid
    if values[mid]<target:return binary_search(values,target,mid+1,hi)
    return binary_search(values,target,lo,mid-1)
print(binary_search([2,4,6,8,10,12],10))
```

## Eg - 9: Flatten arbitrary nested Lists


```python
def flatten(items):
    result=[]
    for item in items:
        if isinstance(item,list):result.extend(flatten(item))
        else:result.append(item)
    return result
print(flatten([1,[2,[3,4]],5,[6]]))
```

## Eg - 10: Complex nested category sales tree

```python
tree={"name":"All","sales":100,"children":[{"name":"Electronics","sales":200,"children":[{"name":"Phones","sales":300,"children":[]}]},{"name":"Books","sales":150,"children":[]}]}
def total_sales(node):
    return node["sales"]+sum(total_sales(child) for child in node.get("children",[]))
print("Total sales:",total_sales(tree))
```
