```python
print("\nQuestion 42: Convert 'hello world' to Morse code")
 
txt = "hello world"

morse_dict = {
    'a': '.-', 'b': '-...', 'c': '-.-.', 'd': '-..',
    'e': '.',  'f': '..-.', 'g': '--.',  'h': '....',
    'i': '..', 'j': '.---', 'k': '-.-',  'l': '.-..',
    'm': '--', 'n': '-.',   'o': '---',  'p': '.--.',
    'q': '--.-','r': '.-.', 's': '...',  't': '-',
    'u': '..-', 'v': '...-', 'w': '.--', 'x': '-..-',
    'y': '-.--','z': '--..',
    ' ': '/'  
}

morse_code = ""
for ch in txt:
    if ch in morse_dict:
        morse_code += morse_dict[ch] + " "

print("Morse code:", morse_code.strip())
```

```python
print("\nQuestion 43: Find the longest common substring between 'programming' and 'grammar'")

str1 = "programming"
str2 = "grammar"

longest = ""
for i in range(len(str1)):
    for j in range(len(str2)):
        temp = ""
        k = 0
       
        while i + k < len(str1) and j + k < len(str2) and str1[i + k] == str2[j + k]:
            temp += str1[i + k]
            k += 1
        if len(temp) > len(longest):
            longest = temp

print("Longest common substring:", longest)
```

```python
print("\nQuestion 46: Convert 'hello world' to Pig Latin")
txt = "hello world"
words = txt.split()
piglatin_words = []

for w in words:
    vowels = "aeiou"
    if w[0].lower() in vowels:
        piglatin_words.append(w + "way")
    else:
        i = 0
        while i < len(w) and w[i].lower() not in vowels:
            i += 1
        if i < len(w):
            piglatin_words.append(w[i:] + w[:i] + "ay")
        else:
            piglatin_words.append(w + "ay")

piglatin = " ".join(piglatin_words)
print("Pig Latin:", piglatin)
```

```python
print("\nQuestion 4: Check if string is a valid IPv4 address: '192.168.1.1'")
txt = "192.168.1.1"

parts = txt.split(".")
valid = True


if len(parts) == 4:
    for p in parts:
        
        if p.isdigit():
            num = int(p)
           
            if num < 0 or num > 255:
                valid = False
                break
        else:
            valid = False
            break
else:
    valid = False

if valid:
    print("Valid IPv4 address")
else:
    print("Invalid IPv4 address")
```

```python
print("\nQuestion 49: Convert 'hello world' to ROT13 encoding")
txt = "hello world"
rot13 = ""

for ch in txt:
    if 'a' <= ch <= 'z':  
        rot13 += chr((ord(ch) - ord('a') + 13) % 26 + ord('a'))
    elif 'A' <= ch <= 'Z':  
        rot13 += chr((ord(ch) - ord('A') + 13) % 26 + ord('A'))
    else:
        rot13 += ch  
print("ROT13:", rot13)
```

```python
print("\nQuestion 50: Check if string is a valid credit card number: '4532015112830366'")

txt = "4532015112830366"

digits = [int(ch) for ch in txt]
checksum = 0

for i in range(len(digits)-1, -1, -1):
    d = digits[i]
    if (len(digits) - i) % 2 == 0:
        d = d * 2
        if d > 9:
            d -= 9
    checksum += d

if checksum % 10 == 0:
    print("Valid credit card number")
else:
    print("Invalid credit card number")
```
