Example - 1: Basic function

```python
def greet():
    return "Welcome to python!"
print(greet())
```

## Example 2 : Parameters and return

```python
def add(a,b):
    return a+b
print(add(12,8))
```

## Example 3: Default arguments

```python
def bill(price, tax =0.18):
    return round(price*(1+tax),2)
print(bill(1000))
```

## Example - 4 : Keyword arguments

```python
def employee(name, role, city):
    return f"{name} works as {role} in {city}"
print(employee(city = "Bengaluru", name = "Asha", role = "Engineer"))
```

## Example - 5: Variable arguments

```python
def total(*numbers):
    return sum(numbers)
print(total(10,20,30,40))
```

## Example 6: Keyword variable arguments

```python
def profile(**details):
    return details
print(profile(name="Ravi", skill ="Python", experience = 3))
```

## Example 7: Multiple Return values


```python
def stats(nums):
    return min(nums), max(nums), sum(nums) / len(nums)
print(stats([10,20,30,40]))
```

## Example - 8: Lambda Functions

```python
square = lambda x:x*x
print([square(x) for x in range(1,6)])
```

## Example - 9: Function as argument


```python
def apply(fn, values):
    return [fn(v) for v in values]
print(apply(lambda x:x*3,[1,2,3]))
```

## Example - 10: Practical Retail discount

```python
def final_price(amount, customer_type ="reular"):
    rates = {"regular":0.05,"silver":0.10,"gold":0.20}
    return round(amount*(1-rates.get(customer_type,0)),2)
orders = [("A",1500,"gold"),("B",850,"regukar"),("c",2200,"silver")]
print([(order_id, final_price(amount,tier)) for order_id, amount, tier in orders])
```
